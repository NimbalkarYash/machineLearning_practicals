```python
import pandas as pd
import seaborn as sns
```

```python
def derivative(x):
    return 2*(x+3)
```

```python
gd = []
def gradient_descent(starting_point, learning_rate, num_iterator):
    x = starting_point
    gd.append(x)
    for i in range(num_iterator):
        grad = derivative(x)
        x = x - learning_rate*grad
        gd.append(x)
    return x
```

```python
start = 2
learning_rate = 0.1
num_iterator = 50

min_x = gradient_descent(start, learning_rate, num_iterator)

min_x
```

```python
sns.scatterplot(gd)
```
