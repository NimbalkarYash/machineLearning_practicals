```python
import pandas as pd
import numpy as nm
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report
```

```python
df = pd.read_csv('2-emails.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace = True)
```

```python
df.head()
```

```python
df.drop("Email No.", axis = 1, inplace = True)

df.head()
```

```python
model1 = KNeighborsClassifier()
```

```python
x = df.drop("Prediction", axis = 1)

x.head()
```

```python
y = df['Prediction']
y.head()
```

```python
from sklearn.model_selection import train_test_split

xtrain, xtest, ytrain, ytest = train_test_split(x,y, test_size= (0.2), random_state = 42)
```

```python
model1.fit(xtrain, ytrain)
model1.score(xtest,ytest)
```

```python
ypreds = model1.predict(xtest)
```

```python
confusion_matrix(ytest, ypreds)
```

```python
classification_report(ytest, ypreds)
```

```python
from sklearn.svm import SVC

model2 = SVC()

model2.fit(xtrain, ytrain)
model2.score(xtest, ytest)
```

```python
ypreds1 = model2.predict(xtest)
```

```python
confusion_matrix(ytest, ypreds1)
```

```python
classification_report(ytest, ypreds1)
```
